```python
import cv2
from src.computer_vision.computer_vision import ComputerVision  # assuming your class is saved as vision.py
from src.computer_vision.vision_params_manager import VisionParamsManager

# 1. Create Vision object
params = VisionParamsManager()
vision = ComputerVision(params)

# 2. Load a test image where the robot is visible
# Load an image/frame for calibration
frame = cv2.imread("vision_live_output_pour_mehdi.jpg")

# 3. Automatically initialize colors and areas
vision.auto_init_colors(frame)

vision.auto_init_canny(frame)

edges = cv2.Canny(frame, vision.params.canny_params["low"], vision.params.canny_params["high"])
cv2.imshow("Canny Edges", edges)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
